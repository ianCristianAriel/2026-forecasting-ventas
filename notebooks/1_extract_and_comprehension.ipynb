#### Diccionario de datos – train.csv y test.csv

Estos archivos contienen información diaria de ventas y actividad comercial de las tiendas.

| Columna | Tipo de columna | Tipo de dato | Descripción | Variables únicas |
|---|---|---|---|---|
| Store | Identificador | Integer (int64) | Identificador único de cada tienda. | 1 a 1115 |
| DayOfWeek | Temporal / Categórica | Integer (int64) | Día de la semana en el que se registra la observación. | 1, 2, 3, 4, 5, 6, 7 |
| Date | Temporal | DateTime | Fecha correspondiente al registro de ventas. | Fechas entre 2013 y 2015 |
| Sales | Numérica continua | Integer (int64) | Importe total de ventas diarias de la tienda. Variable objetivo (Target). | Valores numéricos ≥ 0 |
| Customers | Numérica discreta | Integer (int64) | Cantidad de clientes que visitaron la tienda durante el día. | Valores numéricos ≥ 0 |
| Open | Binaria | Integer (int64) | Indica si la tienda estuvo abierta o cerrada. | 0 = Cerrada, 1 = Abierta |
| Promo | Binaria | Integer (int64) | Indica si la tienda tenía una promoción activa durante el día. | 0 = No, 1 = Sí |
| StateHoliday | Categórica nominal | Object (string) | Identifica si el día corresponde a un feriado estatal. | 0 = Ninguno, a = Feriado público, b = Pascua, c = Navidad |
| SchoolHoliday | Binaria | Integer (int64) | Indica si la observación estuvo afectada por el cierre de escuelas públicas. | 0 = No, 1 = Sí |
| Id | Identificador | Integer (int64) | Identificador único de la combinación entre tienda y fecha. Disponible en test.csv. | Identificadores únicos |

#### Diccionario de datos – store.csv

Este archivo contiene información complementaria sobre las características de cada establecimiento.

| Columna | Tipo de columna | Tipo de dato | Descripción | Variables únicas |
|---|---|---|---|---|
| Store | Identificador | Integer (int64) | Identificador único de cada establecimiento. Permite relacionar este archivo con train.csv y test.csv. | 1 a 1115 |
| StoreType | Categórica nominal | Object (string) | Clasificación del modelo de tienda. | a, b, c, d |
| Assortment | Categórica ordinal | Object (string) | Nivel de variedad de productos ofrecidos por la tienda. | a = Básico, b = Extra, c = Extendido |
| CompetitionDistance | Numérica continua | Float64 | Distancia en metros hasta la tienda competidora más cercana. | Valores numéricos positivos y nulos |
| CompetitionOpenSinceMonth | Temporal / Discreta | Float64 | Mes aproximado de apertura del competidor más cercano. | 1 a 12 |
| CompetitionOpenSinceYear | Temporal / Discreta | Float64 | Año aproximado de apertura del competidor más cercano. | Años entre 1900 y 2015 |
| Promo2 | Binaria | Integer (int64) | Indica si la tienda participa en una promoción continua de largo plazo. | 0 = No participa, 1 = Participa |
| Promo2SinceWeek | Temporal / Discreta | Float64 | Semana del año en la que comenzó la participación de la tienda en Promo2. | 1 a 50 |
| Promo2SinceYear | Temporal / Discreta | Float64 | Año en el que comenzó la participación de la tienda en Promo2. | Años entre 2009 y 2015 |
| PromoInterval | Categórica nominal | Object (string) | Meses en los que se renuevan las promociones de Promo2. | Combinaciones de meses y valores nulos |
